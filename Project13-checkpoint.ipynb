### 🧹 4. Data Cleaning
```markdown
# 🧹 Data tozalash (Data Cleaning)
Bu bosqichda:
- ❗ Missing value (NaN) lar aniqlanadi
- 🔄 Ular mos ravishda to‘ldiriladi

```python
# NaN larni tekshirish
df.isnull().sum()

# Categoric ustunlar uchun mode
for col in df.select_dtypes(include='object'):
    df[col].fillna(df[col].mode()[0], inplace=True)

# Numeric ustunlar uchun median
for col in df.select_dtypes(include='number'):
    df[col].fillna(df[col].median(), inplace=True)

````python
def find_nan_columns():
    columns = df.columns
    nan_columns = []
    for col in columns:
        if df[col].isna().sum() > 0:
            nan_columns.append(col)
    return nan_columns
nan_columns = find_nan_columns()
nan_columns
````

````python
def fill_nan_values(df, nan_columns):
    for col in nan_columns:
        if df[col].dtype == np.dtype('O'):
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].median())
    return df

df = fill_nan_values(df, nan_columns)
````

### 📊 5. Data Visualization
```markdown
# 📊 Data Visualization
Bu bo‘limda datasetni grafiklar orqali tahlil qilamiz.
## 🥧 Gender Distribution
Pie chart va Bar chart orqali visualization

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
female = df["Gender"].value_counts()["female"]
male   = df["Gender"].value_counts()["male"]
values = [female, male]
labels = ["Female", "Male"]
ax1.pie(values,
        labels=labels,
        autopct='%1.1f%%',
        startangle=90,
        explode=(0.02,0),
        colors=["cadetblue","plum"],
        wedgeprops={'edgecolor': 'white', 'linewidth': 1.2})
ax1.set_title("Gender Distribution (Pie)")

gender_counts = df["Gender"].value_counts()
ax2.bar(gender_counts.index,
        gender_counts.values,
        color=["#ff9999", "#66b3ff"],
        edgecolor='white')
ax2.set_title("Gender Distribution (Bar)")
ax2.set_xlabel("Gender")
ax2.set_ylabel("Count")

plt.suptitle("Gender Analysis", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 🥧 Parent Education Distribution
Lollipop chart orqali visualization

````python
counts = df["ParentEduc"].value_counts()

plt.hlines(counts.index, 0, counts.values, color='red', linewidth=1.5)
plt.plot(counts.values, counts.index, "o", color='blue', markersize=8)
plt.xlabel("Count")
plt.title("Parent Education")
plt.tight_layout()
plt.grid(True)
plt.show()
````

## 🥧 Ethnic Distribution
Pie chart va Bar chart orqali visualization

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
gr_C = df["EthnicGroup"].value_counts()["group C"]
gr_D = df["EthnicGroup"].value_counts()["group D"]
gr_B = df["EthnicGroup"].value_counts()["group B"]
gr_E = df["EthnicGroup"].value_counts()["group E"]
gr_A = df["EthnicGroup"].value_counts()["group A"]
ethnic_values = [gr_C,gr_D,gr_B,gr_E,gr_A]
ethnic_labels = ["group C","group D","group B","group E","group A"]
colors = plt.cm.Set3.colors
explode = [0.05 if val == max(ethnic_values) else 0 for val in ethnic_values]
ax1.pie(ethnic_values,
    labels=ethnic_labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=140,
    explode=explode,
    shadow=True,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.2})
ax1.set_title("Ethnic Distribution (Pie)")

ethnic_counts = df["EthnicGroup"].value_counts()
ax2.bar(ethnic_counts.index,
        ethnic_counts.values,
        color=colors,
        edgecolor='black')
ax2.set_title("Ethnic Distribution (Bar)")
ax2.set_xlabel("Ethnic")
ax2.set_ylabel("Count")

plt.title("Ethnic Distribution")
plt.show()
````

````python
from pywaffle import Waffle

counts = df["LunchType"].value_counts()
plt.figure(
    FigureClass=Waffle,
    rows=10,
    values=counts.values.tolist(),
    labels=[f"{k} ({v})" for k, v in zip(counts.index, counts.values)],
    colors=['#66b3ff','#ff9999'],
    legend={'loc': 'lower left', 'bbox_to_anchor': (0, -0.3)})
plt.title("Lunch Type")
plt.show()
````

## 🥧 Scores Distribution

````python
import squarify
counts = df["ParentMaritalStatus"].value_counts()
colors = ["#6baed6", "#74c476", "#fd8d3c", "#9e9ac8"]

fig, axes = plt.subplots(1, 2, figsize=(12,5))

plt.sca(axes[0])

squarify.plot(
    sizes=counts.values,
    label=[f"{k}\n{v}" for k, v in zip(counts.index, counts.values)],
    color=colors,
    alpha=0.85,
    text_kwargs={'fontsize':12,
                 'fontweight':'bold',
                 'color':'grey'}
)

plt.axis("off")
axes[0].set_title("Parent Marital Status (Treemap)")

axes[1].barh(counts.index,
             counts.values,
             color=colors,
             edgecolor="grey",
             linewidth=1.5,
             alpha=0.85)

for i, v in enumerate(counts.values):
    axes[1].text(v + 5, i,
                 str(v),
                 va="center",
                 fontsize=11,
                 fontweight="bold",
                 color="dimgray")

axes[1].set_title("Parent Marital Status (Bar Chart)")
axes[1].grid(axis="x", linestyle="--", alpha=0.3)


plt.tight_layout()
plt.show()
````

## 🥧 Weekly Study Hours Distribution
Subplots Bar chart orqali visualization

````python
order = ["< 5", "5 - 10", "> 10"]
counts = df["WklyStudyHours"].value_counts().reindex(order)

colors = ["#ff9999", "#66b3ff", "#99ff99"]

plt.figure(figsize=(7,5))

bars = plt.bar(order,
               counts.values,
               color=colors,
               edgecolor="grey",
               linewidth=2,
               alpha=0.85)

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 20,
             f"{int(height)}",
             ha="center",
             fontsize=12,
             fontweight="bold",
             color="dimgray")

plt.title("Weekly Study Hours",
          fontsize=15,
          fontweight="bold",
          color="dimgray")

plt.xlabel("Study Hours Category")
plt.ylabel("Number of Students")

plt.grid(axis="y",
         linestyle="--",
         alpha=0.3)

plt.gca().set_facecolor("#f9f9f9")
plt.gcf().patch.set_facecolor("#f9f9f9")

plt.tight_layout()
plt.show()
````

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].hist(df["MathScore"], bins=30, 
             color="#66b3ff", edgecolor="white", linewidth=0.8)
axes[0].set_title("Histogram", fontweight="bold")
axes[0].set_xlabel("Math Score")
axes[0].set_ylabel("Count")

sns.kdeplot(data=df, x="MathScore", 
            fill=True, alpha=0.5, 
            color="#ff9999", ax=axes[1])
axes[1].set_title("KDE Plot", fontweight="bold")
axes[1].set_xlabel("Math Score")

sns.boxplot(data=df, y="MathScore", 
            color="#99ff99", 
            width=0.4, ax=axes[2])
axes[2].set_title("Boxplot", fontweight="bold")

plt.suptitle("Math Score Distribution", 
             fontsize=15, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
````
